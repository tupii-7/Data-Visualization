```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Membuat data untuk variabel x dan y
x = np.linspace(0, 10, 100)
y = np.sin(x)

# --- Cara 1: Menggunakan State-based / Implicit Interface (plt) ---
plt.plot(x, y)
plt.title("(Cara 1 - State-based)")
plt.show()
```

```python
import pandas as pd

# URL digabung dalam satu baris string utuh
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"
df = pd.read_csv(url)

print("Ukuran awal dataset:", df.shape)
print("Jumlah data kosong pada horsepower sebelum diisi:", df["horsepower"].isna().sum())

# Mengisi nilai kosong (NaN) pada kolom horsepower dengan median
df["horsepower"] = df["horsepower"].fillna(df["horsepower"].median())

print("Jumlah data kosong pada horsepower setelah diisi:", df["horsepower"].isna().sum())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 2. Menghitung rata-rata MPG per tahun model
mpg_per_tahun = df.groupby("model_year")["mpg"].mean()

# 3. Membuat grafik (diperbaiki susunan baris kodenya)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(mpg_per_tahun.index, mpg_per_tahun.values, marker="o")
ax.set_title("Rata-rata MPG per Tahun Model")
ax.set_xlabel("Tahun Model (19xx)")
ax.set_ylabel("Rata-rata MPG")

plt.show()
```

```python
mpg_per_asal = (df.groupby("origin")["mpg"]
.mean().sort_values(ascending=False))

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(mpg_per_asal.index,

mpg_per_asal.values, color="steelblue")
ax.set_title("Rata-rata MPG per Asal Mobil")
ax.set_xlabel("Asal Negara")
ax.set_ylabel("Rata-rata MPG")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["horsepower"], df["mpg"],
alpha=0.6)
ax.set_title("Tenaga Mesin vs Efisiensi BBM")
ax.set_xlabel("Horsepower (HP)")
ax.set_ylabel("MPG")
plt.show()

print(df["horsepower"].corr(df["mpg"]))
```

```python
import matplotlib.pyplot as plt

# Mencari mobil paling irit (MPG tertinggi) dan paling boros (MPG terendah)
mobil_teririt = df.loc[df['mpg'].idxmax()]
mobil_terboros = df.loc[df['mpg'].idxmin()]

# Membuat scatter plot sebaran MPG
fig, ax = plt.subplots(figsize=(10, 6))

# Scatter plot untuk semua mobil
ax.scatter(df.index, df['mpg'], color='lightgray', alpha=0.7, label='Mobil Lainnya')

# Highlight mobil paling irit (warna hijau)
ax.scatter(df['mpg'].idxmax(), mobil_teririt['mpg'], color='green', s=150, edgecolors='black', linewidths=2,
           label=f"Paling Irit: {mobil_teririt['name']} ({mobil_teririt['mpg']} MPG)")

# Highlight mobil paling boros (warna merah)
ax.scatter(df['mpg'].idxmin(), mobil_terboros['mpg'], color='red', s=150, edgecolors='black', linewidths=2,
           label=f"Paling Boros: {mobil_terboros['name']} ({mobil_terboros['mpg']} MPG)")

# Pengaturan label dan judul grafik
ax.set_title("Sebaran Data MPG Mobil (Highlight Paling Irit & Boros)", fontsize=14)
ax.set_xlabel("Indeks Data / Mobil")
ax.set_ylabel("Miles Per Gallon (MPG)")
ax.legend(loc='upper right')
ax.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(df["weight"], df["mpg"], alpha=0.6, color="purple")
ax.set_title("Hubungan Bobot Mobil (Weight) vs Efisiensi (MPG)")
ax.set_xlabel("Bobot Mobil (Weight)")
ax.set_ylabel("MPG")
ax.grid(True, alpha=0.3)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 6))
# Ukuran bubble disesuaikan dengan nilai acceleration (dikali skalar agar terlihat)
scatter = ax.scatter(
    df["horsepower"],
    df["mpg"],
    s=df["acceleration"] * 10,
    c=df["weight"],
    cmap="viridis",
    alpha=0.6
)

ax.set_title("Horsepower vs MPG (Ukuran = Acceleration, Warna = Weight)")
ax.set_xlabel("Horsepower")
ax.set_ylabel("MPG")
fig.colorbar(scatter, ax=ax, label="Weight")
ax.grid(True, alpha=0.3)
plt.show()
```

```python
jumlah = df["origin"].value_counts()

fig, ax = plt.subplots(figsize=(6, 3))
ax.barh(jumlah.index, jumlah.values, color="green")
ax.set_title("Jumlah Mobil per Asal")
ax.set_xlabel("Jumlah Mobil")
ax.set_ylabel("Asal Mobil")

plt.show()
```

```python
# Menghitung rata-rata weight per tahun model
weight_per_tahun = df.groupby("model_year")["weight"].mean()

# Membuat Line Chart
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(weight_per_tahun.index, weight_per_tahun.values, marker="o", color="violet", linewidth=2)
ax.set_title("Rata-rata Bobot Mobil (Weight) per Tahun Model")
ax.set_xlabel("Tahun Model (19xx)")
ax.set_ylabel("Rata-rata Bobot (lbs)")
ax.grid(True, linestyle="--", alpha=0.7)

plt.show()
```
