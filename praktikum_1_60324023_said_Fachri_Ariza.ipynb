### Conditional Statements (if-elif-else)

```python
score = 85

if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
else:
    print("Grade D")
```

### Function Definition (def)

```python
def hitung_rata2(data):
  total = sum(data)
  jumlah = len(data)
  rata2 = total / jumlah
  return rata2

nilai = [90, 87, 76, 98, 82]
hasil = hitung_rata2(nilai)
print ("rata rata nilainya",hasil)
```

```python
nama = "said Fachri Ariza"
nilai = 99

def kategori_nilai(nilai):
    if nilai >= 90:
        return "sangat baik"
    elif nilai >= 80:
        return "Baik"
    elif nilai >= 70:
        return "cukup"
    else:
        return "perlu belajar lagi"

print(f"nama: {nama}")
print(f"nilai: {nilai}")
print(f"kategori nilai: {kategori_nilai(nilai)}")
```
