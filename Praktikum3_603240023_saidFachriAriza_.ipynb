```python
import numpy as np
import pandas as pd
```

```python
# Membuat array NumPy dari sebuah list
nilai_ujian = np.array([78, 85, 90, 65, 72])
print("Rata-rata:", np.mean(nilai_ujian))
print("Maksimum:", np.max(nilai_ujian))
print("Minimum:", np.min(nilai_ujian))
print("Std Deviasi:", np.std(nilai_ujian))
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)
df.head() # menampilkan 5 baris pertama
```

```python
print(df.shape) # jumlah baris & kolom
print(df.columns) # nama-nama kolom
df.info() # tipe data & jumlah data tidak kosong
df.describe() # statistik deskriptif kolom numerik
```

```python
# Mengakses satu kolom
df["age"].head()
# Mengakses beberapa kolom sekaligus
df[["age", "sex", "survived"]].head()
# Memfilter baris berdasarkan kondisi
penumpang_anak = df[df["age"] < 18]
print("Jumlah penumpang anak:", len(penumpang_anak))
```

```python
print(df.isnull().sum())
```

```python
# Strategi 1: isi nilai kosong dengan rata-rata (untuk data numerik)
df["age"] = df["age"].fillna(df["age"].mean())
# Strategi 2: hapus baris yang datanya kosong pada kolom tertentu
df = df.dropna(subset=["embarked"])
print(df.isnull().sum()) # periksa kembali
```

```python
print("Jumlah data sebelum:", len(df))
print("Jumlah duplikasi:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Jumlah data sesudah:", len(df))
```

```python
df.to_csv("titanic_bersih.csv", index=False)
```

# latihan mandiri

```python
# 10 baris pertama
print("--- 10 Baris Pertama ---")
print(df.head(10))

# 5 baris terakhir
print("\n--- 5 Baris Terakhir ---")
print(df.tail(5))
```

```python
# Menghitung dengan NumPy
mean_numpy = np.mean(df['age'])

# Menghitung dengan Pandas
mean_pandas = df['age'].mean()

print(f"Rata-rata Usia (NumPy)  : {mean_numpy}")
print(f"Rata-rata Usia (Pandas) : {mean_pandas}")

# Membandingkan hasil
if np.isclose(mean_numpy, mean_pandas, equal_nan=True):
    print("Hasilnya SAMA.")
else:
    print("Hasilnya BEDA.")
```

```python
# Filter kondisi ganda: pclass 1 DAN survived 1
penumpang_kelas1_selamat = df[(df['pclass'] == 1) & (df['survived'] == 1)]

# Menampilkan data
print("Data Penumpang Kelas 1 yang Selamat:")
print(penumpang_kelas1_selamat.head())

print("\nJumlah penumpang kelas 1 yang selamat:", len(penumpang_kelas1_selamat))
```

```python
# Mencari modus kolom embarked
modus_embarked = df['embarked'].mode()[0]
print("Modus kolom embarked:", modus_embarked)

# Mengisi missing value dengan modus
df['embarked'] = df['embarked'].fillna(modus_embarked)

# Cek kembali jumlah missing value pada kolom embarked
print("Sisa missing value kolom embarked:", df['embarked'].isnull().sum())
```

```python
# Menghitung persentase
persentase_selamat = df['survived'].mean() * 100

print(f"Persentase penumpang yang selamat: {persentase_selamat:.2f}%")
```
