```python
def hitung_rata2(data):
  """Menghitung nilai rata-rata dari sebuah list."""
  total = sum(data)
  jumlah = len(data)
  return total / jumlah
# Memanggil fungsi
nilai_kelas = [80, 75, 90, 65, 88]
hasil = hitung_rata2(nilai_kelas)

print("Rata-rata kelas:", hasil)
```

```python
def kategori_nilai(n):
  if n >= 85:
    return "Sangat Baik"
  elif n >= 70:
    return "Baik"
  else:
    return "Perlu Belajar Lagi"
# Terapkan ke seluruh nilai kelas
for n in nilai_kelas:
  print(n, "->", kategori_nilai(n))
```

```python
def nilai_tertinggi(data):
    maks = data[0]

    # Cek setiap nilai dalam list
    for n in data:
        if n > maks:
            maks = n # Perbarui nilai tertinggi jika ditemukan angka yang lebih besar

    return maks

nilai_kelas = [80, 75, 90, 65, 88]
print("Nilai tertinggi:", nilai_tertinggi(nilai_kelas))
```

```python
def kategori_nilai(n):
    if n >= 85:
        return "Sangat Baik"
    elif n >= 70:
        return "Baik"
    elif n >= 60:
        return "Cukup"
    else:
        return "Kurang"

# Pengujian dengan data nilai_kelas
nilai_kelas = [80, 75, 90, 65, 88, 55]

for n in nilai_kelas:
    print(n, "->", kategori_nilai(n))
```

```python
# List berisi 8 nilai ujian mahasiswa
nilai_ujian = [85, 62, 78, 90, 55, 70, 82, 48]

# Inisialisasi variabel penghitung
lulus = 0
tidak_lulus = 0

# Perulangan untuk memeriksa setiap nilai
for nilai in nilai_ujian:
    if nilai >= 70:
        lulus += 1         # Tambah 1 ke variabel lulus jika nilai >= 70
    else:
        tidak_lulus += 1   # Tambah 1 ke variabel tidak_lulus jika nilai < 70

# Menampilkan hasil
print("Daftar Nilai Ujian:", nilai_ujian)
print("Jumlah mahasiswa LULUS (≥ 70):", lulus)
print("Jumlah mahasiswa TIDAK LULUS (< 70):", tidak_lulus)
```

```python
def konversi_nilai(n):
    """Mengubah nilai angka (0-100) menjadi huruf grade A/B/C/D."""
    if n >= 85:
        return "A"
    elif n >= 70:
        return "B"
    elif n >= 60:
        return "C"
    else:
        return "D"

# List 8 nilai ujian dari Latihan 4
nilai_ujian = [85, 62, 78, 90, 55, 70, 82, 48]

# Menerapkan fungsi ke seluruh elemen list menggunakan perulangan
print("--- Hasil Konversi Nilai Ujian ---")
for nilai in nilai_ujian:
    grade = konversi_nilai(nilai)
    print(f"Nilai: {nilai} -> Grade: {grade}")
```
